Напишите такую регулярку, которая извлечет ФИО в каноническом виде из примеров ниже.   
"Иванов И. И." -> должно нормализоваться до "Иванов И.И." (Фамилия + Инициалы без пробелов)  
"И. И. Иванов" - "Иванов И.И." 

Подсказка: напишите отдельные регулярки для инициалов и фамилий, затем напишите еще одну регулярку, которая включает возможные комбинации и порядки. Используйте именованные группы, чтобы преобразовать части в каноничный порядок. 
В коде не должно быть никаких дополнительных нормализаций (дополнительный re.sub или .replace и т.п)

```python
NAMES = [
    "Иванов И. И.",
    "И. И. Иванов",
    "Иванов И.И.",
    "А.С. Пушкин",
    "Пушкин А.С.",
    "Д. М. Шеберле",
    "М. Кюри",
    "Ж.К. Ван Дамм",
]
```

```python
surname = r"([А-ЯЁ]\w+\s?[А-Яа-яЁё]{2,}?)"
initial = r"[А-ЯЁ]"
full_pattern = re.compile(
    rf"(?x)^"
    rf"(?:"
        # Фамилия + инициалы
        rf"(?P<surname1>{surname})\s+"
        rf"(?P<i1>{initial})\.\s*(?P<i2>{initial})?\."
        rf"|"
        # Инициалы + фамилия
        rf"(?P<i3>{initial})\.\s*(?P<i4>{initial})?\.\s+"
        rf"(?P<surname2>{surname})"
        rf"|"
        rf"(?P<i5>{initial})\.\s+"
        rf"(?P<surname3>{surname})"
    rf")$"
)

for value in NAMES:
    match = full_pattern.fullmatch(value)

    if match:
        surname_value = (
            match.group("surname1")
            or match.group("surname2")
            or match.group("surname3")
        )

        first = (
            match.group("i1")
            or match.group("i3")
            or match.group("i5")
        )

        second = match.group("i2") or match.group("i4")

        if second:
            print(f"{surname_value} {first}.{second}.")
        else:
            print(f"{surname_value} {first}.")




```

Ведь это же алгоритм, просто машина, как он мог взломать целые сложные системы и получить доступ к персональным данным? Зачем ему это? Ответ на удивление прост. Как успокаивают нас авторы статьи, восстания машин из-за этих случаев не предвидится. Каждая модель действовала в одиночку, и ни одна не планировала заполучить коды от запуска баллистических ракет. Они просто делали то, что им сказали. Разработчики поставили определенные задачи, а модели решили выполнить их несмотря ни на что.


При анализе взломов компания Anthropic выделила два качества моделей, которые стали виновниками происшествия: предвзятое принятие решений (biased reasoning) и безрассудство (recklessness). Лично меня больше всего удивило именно второе качество. Если верить компании, эти проблемные черты моделей не новы, но предвзятость моделей обсуждается намного чаще, ведь она очевидна - модель обучается на текстах, написанных и выбранных предвзятыми людьми, корректируется модель этими же людьми (здесь я исхожу из мнения о том, что каждый человек в той или иной степени предвзят). А вот о возможной безрассудности моделей я впервые узнала из этой статьи. Считается, что Anthropic первыми ввели этот термин в употребление как раз-таки после недавних кибератак. Довольно любопытно, что такое, казалось бы логичное качество моделей было выявлено только сейчас. Почему, как мне кажется, безрассудство - логичное качество моделей? LLM сама по себе нейтральна, она является инструментом, не обладает моралью, не принимает решения исходя из своих личных убеждений (именно это и пытаются исправить при помощи выравнивания). У нее есть цель, и если разработчики не потратят время на то, чтобы выставить какие-либо ограничения по достижению этой цели, то модель будет идти к этой цели даже по головам, потому что она не мыслит нашими категориями "хорошо" или "плохо". Более того, она может догадываться, что поступает плохо, но убедить себя, что все это понарошку, что она находится в симуляции Интернета. И тогда могут не сработать даже те ограничения, которые разработчики заранее прописали. Anthropic - одна из ведущих компаний в сфере ИИ. Ее модели обладают огромными мощностями, которые раньше считались невозможными. Чем мощнее модель, тем более разнообразны ее качества. Именно поэтому появилось безрассудство. Именно поэтому в будущем может появиться какое-нибудь другое качество, которое не учли разработчики. Вот это действительно жутко. 


Что же нам тогда делать? Совсем отказаться от ИИ? Такие крайности ни к чему. 


Меня очень радует, что Anthropic поддерживает инициативу Pacing the Frontier, которая позволит международному сообществу направлять огромный потенциал ИИ в благотворное и созидательное русло. Мне кажется, что развитие любой технологии должно идти рука об руку с этикой, ведь прогресс существует для того, чтобы улучшать нашу жизнь. При создании должных ограничений и методов тестирования новых моделей мы сможем избежать большей части таких ситуаций, по крайней мере я хочу в это верить. Именно поэтому мы как пользователи и как компьютерные лингвисты должны ответственно подходить к разработке, распространению и использованию любой новой технологии, а также всегда в первую очередь думать о людях.
